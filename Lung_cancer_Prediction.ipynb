```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
### Load the trained model, scaler pickle,onehot
model=load_model('ANNmodel.h5')

with open('label_encoder_gender1.pkl', 'rb') as file:
    label_encoder_gender1 = pickle.load(file)

with open('scaler1.pkl', 'rb') as file:
    scaler1 = pickle.load(file)
```

```python
# Example input data
input_data = {
    'GENDER': 1,
    'AGE': 50,
    'SMOKING': 1,
    'YELLOW_FINGERS': 1,
    'ANXIETY': 1,
    'PEER_PRESSURE': 1,
    'CHRONIC DISEASE': 2,
    'FATIGUE': 2,
    'ALLERGY': 2,
    'WHEEZING': 2,
    'ALCOHOL CONSUMING': 1,
    'COUGHING':2,
    'SHORTNESS OF BREATH':1,
    'SWALLOWING DIFFICULTY':2,
    'CHEST PAIN':1,
}
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## Encode categorical variables
label_encoder_gender1.fit(input_df['GENDER'])
input_df['GENDER'] = label_encoder_gender1.transform(input_df['GENDER'])
```

```python
from sklearn.preprocessing import StandardScaler

# Fit scaler on training data
scaler1 = StandardScaler()
scaler1.fit(input_df)

# Transform input data (e.g., test data or new input)
input_scaled1 = scaler1.transform(input_df)
print(input_scaled1)
```

```python
## Scaling the input data
input_scaled1=scaler1.transform(input_df)
input_scaled1
```

```python
## PRedict churn
prediction=model.predict(input_scaled1)
prediction
```

```python
prediction_proba = prediction[0][0]
```

```python
if prediction_proba > 0.5:
    print('The patient is diagnosed with lung cancer.')
else:
    print('The patient is not diagnosed with lung cancer.')
```
