```python
#importing libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
## Load the dataset
Dataset=pd.read_csv("lung cancer.csv")
Dataset.head()
```

```python
## Encode categorical variables gender as 0 and 1
label_encoder_gender=LabelEncoder()
Dataset['GENDER']=label_encoder_gender.fit_transform(Dataset['GENDER'])
Dataset
```

```python
## Encode categorical variables gender as 0 and 1
label_encoder_gender=LabelEncoder()
Dataset['LUNG_CANCER']=label_encoder_gender.fit_transform(Dataset['LUNG_CANCER'])
Dataset
```

```python
## Save the encoders and sscaler
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder_gender,file)
```

```python
Dataset.head()
```

```python
## DiVide the dataset into indepent and dependent features
Independent=Dataset.drop('LUNG_CANCER',axis=1)
Dependent=Dataset['LUNG_CANCER']

## Split the data in training and tetsing sets
X_train,X_test,y_train,y_test=train_test_split(Independent,Dependent,test_size=0.2,random_state=42)

## Scale these features
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```
