```python
#importing libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
## Load the dataset
Dataset=pd.read_csv("lung cancer.csv")
Dataset.head()
```

```python
Dataset.columns
```

```python
## Encode categorical variables gender as 0 and 1
label_encoder_gender1=LabelEncoder()
Dataset['GENDER']=label_encoder_gender1.fit_transform(Dataset['GENDER'])
Dataset
```

```python
## Encode categorical variables gender as 0 and 1
label_encoder_gender=LabelEncoder()
Dataset['LUNG_CANCER']=label_encoder_gender.fit_transform(Dataset['LUNG_CANCER'])
Dataset
```

```python
## Save the encoders and sscaler
with open('label_encoder_gender1.pkl','wb') as file:
    pickle.dump(label_encoder_gender1,file)
```

```python
Dataset.head()
```

```python
## DiVide the dataset into indepent and dependent features
Independent=Dataset.drop('LUNG_CANCER',axis=1)
Dependent=Dataset['LUNG_CANCER']

## Split the data in training and tetsing sets
X_train,X_test,y_train,y_test=train_test_split(Independent,Dependent,test_size=0.2,random_state=42)

## Scale these features
scaler1=StandardScaler()
X_train=scaler1.fit_transform(X_train)
X_test=scaler1.transform(X_test)
```

```python
X_train
```

```python
with open('scaler1.pkl','wb') as file:
    pickle.dump(scaler1,file)
```

ANN Implementaion

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
(X_train.shape[1],)
```

```python
## Build Our ANN Model
model=Sequential([
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)), ## HL1 Connected wwith input layer
    Dense(32,activation='relu'), ## HL2
    Dense(1,activation='sigmoid')  ## output layer
]

)
```

```python
model.summary()
```

```python
import tensorflow
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
loss=tensorflow.keras.losses.BinaryCrossentropy()
loss
```

```python
## compile the model
model.compile(optimizer=opt,loss="binary_crossentropy",metrics=['accuracy'])
```

```python
## Set up the Tensorboard
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard

log_dir="logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
## Set up Early Stopping
early_stopping_callback=EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
```

```python
### Train the model
history=model.fit(
    X_train,y_train,validation_data=(X_test,y_test),epochs=100,
    callbacks=[tensorflow_callback,early_stopping_callback]
)
```

```python
model.save('ANNmodel.h5')
```
